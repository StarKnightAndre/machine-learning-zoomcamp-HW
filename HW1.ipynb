## Q.6)

```python
initial_median = df['horsepower'].median()

most_frequent_value = df['horsepower'].mode()[0]

horsepower_filled = df['horsepower'].fillna(most_frequent_value)

new_median = horsepower_filled.median()

print(f"Initial median: {initial_median}")
print(f"New median: {new_median}")
```

## Q.7)

```python
import numpy as np

asia_cars = df[df['origin'] == 'Asia']
subset = asia_cars[['vehicle_weight', 'model_year']].iloc[:7]

X = subset.values

XTX = X.T.dot(X)

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv.dot(X.T).dot(y)

final_sum = w.sum()

print(final_sum)
```
